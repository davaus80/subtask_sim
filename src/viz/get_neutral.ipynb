## Getting neutral terms
Using this code, we identify words that have neither positive nor negative sentiment according to Saif Mohammed's sentiment lexicon.

```python
neg_path = "/Users/david/Documents/Projects/Skyfall/NRC-Emotion-Lexicon/OneFilePerEmotion/negative-NRC-Emotion-Lexicon.txt"
pos_path = "/Users/david/Documents/Projects/Skyfall/NRC-Emotion-Lexicon/OneFilePerEmotion/positive-NRC-Emotion-Lexicon.txt"
```

```python
def load_scores(path, lower=False):
    d = {}
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith("#"):
                continue
            parts = line.split()
            if len(parts) < 2:
                continue
            word = parts[0]
            if lower:
                word = word.lower()
            # last token should be the numeric score (handles if tabs/spaces vary)
            try:
                score = int(parts[-1])
            except ValueError:
                continue
            d[word] = score
    return d

# use the paths you already defined
neg = load_scores(neg_path, lower=True)
pos = load_scores(pos_path, lower=True)

# words present in both files with score 0 in both
neutral_words = sorted(w for w in (set(neg) & set(pos)) if neg[w] == 0 and pos[w] == 0)

print(f"Neutral words (count={len(neutral_words)}). First 50:")
print(neutral_words[:50])

# optional: save to a file in the repo
out_path = "neutral_words.txt"
with open(out_path, "w", encoding="utf-8") as out:
    out.write("\n".join(neutral_words))

print(f"Saved neutral words to: {out_path}")
```

```python
## Randomly sample 3 neutral words
import random

k = 3  # Change this to sample a different number
sampled = random.sample(neutral_words, k)
print(f"Randomly sampled {k} neutral words:")
print(sampled)
```
