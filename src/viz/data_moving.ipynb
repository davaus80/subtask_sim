```python
# Load in dataframes
for idx, df_metadata in enumerate(df_list):
    # Load
    temp_loaded_df = pd.read_csv(df_metadata['df_path'])
    # Add differentiating columns
    temp_loaded_df['Model'] = df_metadata['model']
    temp_loaded_df['History'] = df_metadata['history']
    temp_loaded_df['Domain'] = df_metadata['domain']
    temp_loaded_df['Variance'] = df_metadata['variance']
    # Save
    df_list[idx]['df'] = temp_loaded_df
```

```python
# Merge the DFs
merged_df = pd.concat([
    df_list_item['df'] for df_list_item in df_list
    ], ignore_index=True)
```
