## Loading Farm NoVar

```python
import pandas as pd

farm_qwen8B_fullhist_df = pd.read_csv("experiments/20260118_farm_novar/model_name_Qwen3-8B_prompt_fullhist/eval_results.csv")
farm_qwen8B_summhist_df = pd.read_csv("experiments/20260118_farm_novar/model_name_Qwen3-8B_prompt_summhist/eval_results.csv")
farm_qwen14B_fullhist_df = pd.read_csv("experiments/20260118_farm_novar/model_name_Qwen3-14B_prompt_fullhist/eval_results.csv")
farm_qwen14B_summhist_df = pd.read_csv("experiments/20260118_farm_novar/model_name_Qwen3-14B_prompt_summhist/eval_results.csv")
```

```python
# Add differentiating columns
farm_qwen8B_fullhist_df['Model'] = "Qwen8B"
farm_qwen8B_fullhist_df['History'] = "Full"
farm_qwen8B_summhist_df['Model'] = "Qwen8B"
farm_qwen8B_summhist_df['History'] = "Summarized"
farm_qwen14B_fullhist_df['Model'] = "Qwen14B"
farm_qwen14B_fullhist_df['History'] = "Full"
farm_qwen14B_summhist_df['Model'] = "Qwen14B"
farm_qwen14B_summhist_df['History'] = "Summarized"
```

```python
# Merge the DFs
merged_df = pd.concat([
    farm_qwen8B_fullhist_df,
    farm_qwen8B_summhist_df,
    farm_qwen14B_fullhist_df,
    farm_qwen14B_summhist_df
], ignore_index=True)
```

```python
merged_df.to_csv("experiments/20260118_farm_novar/merged_eval_results.csv")
```
