# Bandit plotting
The purpose of this notebook is to plot the preliminary results on the simple single-bandit task.

```python
import matplotlib.pyplot as plt
import numpy as np 

from typing import List, Dict, Any, str
```

```python
import jsonlines
## Load in all the relevant results files
def load_results_from_file(jsonl_path: Dict[str, str]) -> Dict[str, List[Dict]]:
    results = []
    with jsonlines.open(jsonl_path) as reader:
        for obj in reader:
            results.append(obj)
    return results
```

```python
def get_reward_only(results_dict: Dict[str, List[Dict]]) -> Dict[str, List[Dict]]:
    cum_reward_dict = {}
    step_reward_dict = {}
    for result_name, results in results_dict.items():
        reward_list = []
        cum_reward_list = []
        cum_reward = 0.0
        # Iterate over the reward at each step, maintaining per-step and cumulative rewards
        for step_idx, step_results in enumerate(results):
            reward_list.append(step_results['reward'])
            cum_reward += step_results['reward']
            cum_reward_list.append(cum_reward)
            
        individual_reward_dict = {
            'reward': reward_list,
            'cum_reward': cum_reward_list
        }
        
        step_reward_dict[result_name] = reward_list
        cum_reward_dict[result_name] = cum_reward_list
    
    return step_reward_dict, cum_reward_dict
```

```python
## Great a line graph where each line is labelled accordingly
def draw_reward_plot(results_dict, title, x_lab="Turn", y_lab="Reward"):
    # Initialize the plot
    plt.figure(figsize=(10, 6))
    
    # We expect results_dict to be keys (used in the legend) and a list of values to plot at each timestep
    for label, rewards in results_dict.items():
        plt.plot(rewards, label=label)  # Plot each reward list with its corresponding label
    
    # Apply title and labels from arguments
    plt.title(title)
    plt.xlabel(x_lab)
    plt.ylabel(y_lab)
    
    # Add a legend to differentiate the lines
    plt.legend()
    
    # Display the plot
    plt.show()
```

```python
## Create a script which calls all of the above. Takes in axes labels, title, and results path dict and outputs a plot
def create_reward_plots(result_path_dict, title, x_lab="Turn", y_lab="Reward"):
    results_dict = {}
    for result_name, result_path in result_path_dict.items():
        results_dict[result_name] = load_results_from_file(result_path)
    
    step_reward_dict, cum_reward_dict = get_reward_only(results_dict)
    
    draw_reward_plot(step_reward_dict, title, x_lab, y_lab)
    
    draw_reward_plot(cum_reward_dict, title, x_lab, "Cumulative Reward")
```
