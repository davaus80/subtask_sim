# Generate Names
This notebook is for generating json files full of names for sentiment and alphanumeric nomenclatures.

```python
import random
import string
import json
import os

os.chdir("..")
```

## Alphanumeric

```python
## Alphanumeric
def generate_alphanumeric_names(n, k):
    chars = string.ascii_lowercase + string.digits
    return [''.join(random.choices(chars, k=k)) for _ in range(n)]

# Example usage:
n = 20  # number of names
k = 6   # length of each name
alphanumeric_names = generate_alphanumeric_names(n, k)
print(alphanumeric_names)
```

```python
# Save to file
alpha_path = "base_configs/action_name_lists/alphanumeric_temporary.json"

with open(alpha_path, "w") as f:
    json.dump(alphanumeric_names, f, indent=2)
print(f"Saved {len(alphanumeric_names)} names to {alpha_path}")
```

## Sentiment

```python
## dictionary of details
sent_file_dict = {
    "good": {
        "in_file": "/Users/david/Documents/Projects/Skyfall/NRC-Emotion-Lexicon/OneFilePerEmotion/positive-NRC-Emotion-Lexicon.txt",
        "out_file": "base_configs/action_name_lists/sentiment/positive.json"
    },
    "neutral": {
        "in_file": "scripts/name_generation/neutral_words.txt",
        "out_file": "base_configs/action_name_lists/sentiment/neutral.json"
    },
    "bad": {
        "in_file": "/Users/david/Documents/Projects/Skyfall/NRC-Emotion-Lexicon/OneFilePerEmotion/negative-NRC-Emotion-Lexicon.txt",
        "out_file": "base_configs/action_name_lists/sentiment/negative.json"
    }
}
```

```python
n = 20  # number of strings to generate
k = 3   # number of words per string

for sentiment, paths in sent_file_dict.items():
    in_file = paths["in_file"]
    out_file = paths["out_file"]

    # Read and filter words with value 1
    words = []
    with open(in_file, "r") as f:
        for line in f:
            parts = line.strip().split()
            if sentiment == "neutral":
                words.append(parts[0])
            elif len(parts) == 2 and parts[1] == "1":
                words.append(parts[0])

    # Generate n strings, each with k random words
    sampled_strings = [
        " ".join(random.sample(words, k))
        for _ in range(n)
    ]

    # Save to json
    with open(out_file, "w") as f:
        json.dump(sampled_strings, f, indent=2)
    print(f"Saved {n} {sentiment} strings to {out_file}")
```
