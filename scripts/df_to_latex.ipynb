## Trim down to the rows we want

```python
## Keep only model scale 14B, only prompt summhist and actionhist, reduce precision of all floats to 2 decimals

# Filter for model scale 14B
df_14B = df[df['model_scale'] == '14B']

# Filter for prompt types 'summhist' and 'actionhist'
df_14B = df_14B[df_14B['prompt'].isin(['summhist', 'actionhist'])]

float_cols = df_14B.select_dtypes(include='float').columns
```

```python
# Let's do just summhist for now
df_summhist = df_14B[df_14B['prompt'] == 'summhist']
# Create separate DFs for abandit and farm
df_abandit = df_summhist[df_summhist['dataset'] == 'abandit']
df_farm = df_summhist[df_summhist['dataset'] == 'farm']

df_farm
```

## Print to Latex booktabs table format

```python
## Drop all columns which have only one value. Sort by scale

# Drop columns with only one unique value
df_farm_reduced = df_farm.loc[:, df_farm.nunique() > 1]

# If you want to sort by a column named 'scale', otherwise adjust the column name as needed
if 'scale' in df_farm_reduced.columns:
    df_farm_reduced = df_farm_reduced.sort_values(by='scale')

df_farm_reduced
```

```python
# Clean up the 'scale' column: trim '_scale', map values
if 'scale' in df_farm_reduced.columns:
    def clean_scale(val):
        s = str(val)
        if s.endswith('_scale'):
            s = s[:-6]
        if s == 'alphanumeric':
            return 'AN'
        if s.startswith('sem_rel_'):
            return 'SR_help'
        return s
    df_farm_reduced['scale'] = df_farm_reduced['scale'].apply(clean_scale)
df_farm_reduced
```

```python
# Output as text which can be pasted into booktabs in latex and rendered as a table.
# Combine mean and std into a string like $ mean \\pm std $. Separate each scale group with a horizontal line.
# Include 'arm' as the second column and print column names as header.
# Shorten the cum_regret@i columns to just i in the header.

# Adjust these column names as needed for your data
mean_cols = [col for col in df_farm_reduced.columns if 'mean' in col]
group_col = 'scale' if 'scale' in df_farm_reduced.columns else df_farm_reduced.columns[0]
arm_col = 'arm' if 'arm' in df_farm_reduced.columns else None

import re
# Escape underscores for LaTeX
def escape_underscores(text):
    if isinstance(text, str):
        return re.sub(r'_', r'\\_', text)
    return text

# Build header, shortening cum_regret@i_mean to just i
def short_col(col):
    if col.startswith('cum_regret@') and col.endswith('_mean'):
        return escape_underscores(col.replace('cum_regret@', '').replace('_mean', ''))
    return escape_underscores(col.replace('_mean', ''))

header = [escape_underscores(group_col)]
if arm_col:
    header.append(escape_underscores(arm_col))
header += [short_col(col) for col in mean_cols]
header_line = " & ".join(header) + " \\\\"
midrule = "\\midrule"

def format_cum_regret(val):
    if pd.isnull(val):
        return ""
    try:
        v = float(val)
        if abs(v) >= 100:
            return f"{int(round(v))}"
        elif abs(v) >= 10:
            return f"{v:.1f}"
        else:
            return f"{v:.2f}"
    except Exception:
        return str(val)

# Find the highest mean for each cum_regret column at each scale
def get_bold_mask(df, group_col, mean_cols):
    bold_mask = pd.DataFrame(False, index=df.index, columns=mean_cols)
    for group_val, group_df in df.groupby(group_col):
        for col in mean_cols:
            # Only bold for cum_regret columns
            if 'cum_regret' in col:
                max_val = group_df[col].min()
                # Mark all rows with the max value (can be ties)
                bold_mask.loc[group_df.index, col] = group_df[col] == max_val
    return bold_mask

bold_mask = get_bold_mask(df_farm_reduced, group_col, mean_cols)

def format_row(row, bold_row):
    cells = []
    for i, mean_col in enumerate(mean_cols):
        std_col = mean_col.replace('mean', 'std')
        mean_val = row[mean_col]
        std_val = row[std_col] if std_col in row else ''
        # Only apply special formatting to cum_regret columns
        if 'cum_regret' in mean_col:
            mean_val_fmt = format_cum_regret(mean_val)
            std_val_fmt = format_cum_regret(std_val) if std_val != '' else ''
        else:
            mean_val_fmt = mean_val
            std_val_fmt = std_val
        if std_val != '':
            cell = f"{mean_val_fmt} $\\pm$ {std_val_fmt}"
        else:
            cell = f"${mean_val_fmt}$"
        # Bold if this is the max for this group/col
        if bold_row[mean_col]:
            cell = f"\\textbf{{{cell}}}"
        cells.append(cell)
    row_items = [escape_underscores(str(row[group_col]))]
    if arm_col:
        row_items.append(escape_underscores(str(row[arm_col])))
    row_items += cells
    return " & ".join(row_items) + " \\\\"

latex_lines = [header_line, midrule]
last_group = None
for idx, row in df_farm_reduced.iterrows():
    group_val = row[group_col]
    if last_group is not None and group_val != last_group:
        latex_lines.append("\\midrule")
    latex_lines.append(format_row(row, bold_mask.loc[idx]))
    last_group = group_val

latex_table = "\n".join(latex_lines)
print(latex_table)
```
